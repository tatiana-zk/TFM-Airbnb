```python
import pandas as pd
import geopandas as gpd
```

```python
lugares = [
    "Plaça Catalunya, Barcelona",
    "Barceloneta, Barcelona",
    "Sagrada Familia, Barcelona",
    "Park Güell, Barcelona",
    "Camp Nou, Barcelona"
]

geo = gpd.tools.geocode(lugares).to_crs(25831)
geo.index = [
    "dist_center_km","dist_beach_km","dist_sagrada_familia_km",
    "dist_park_guell_km","dist_camp_nou_km"
]

geo.to_file("geo_barcelona.geojson", driver="GeoJSON")
```
